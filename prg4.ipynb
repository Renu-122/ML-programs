```python
#using k-fold cross validations and linear regression
'''
steps:
1.divide the 10 records into 5 folds with a 2 records in each fold.
2.perform 5 iterations of k-fold cross validation by keeping one fold as the validation set and the remaining 4 fold as the training data.
3. for each iterations:
  a) calculate the mean x and y of each trainig data.
  b)Calculate the slope and intercept.
'''
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
from numpy import average
```

```python
#load data
# 2. Load dataset
df = pd.read_csv('student_dataset.csv')
print(df)
```

```python
# 3.Separate Input (X) and Output (Y)
x = df[["Hours_Studied"]]
y = df[["Marks"]]
print("x: ")
print(x)

print("\ny: ")
print(y)
```

```python
# 4. Create 5 Folds
kf = KFold(n_splits=5, shuffle=False)
print("Number of folds: ",kf.n_splits)
```

```python
# 5. Perform k-fold cross validation
mse_values = []
fold = 1

for train_index, validation_index in kf.split(x):
  x_train = x.iloc[train_index]
  y_train = y.iloc[train_index]

  x_validation = x.iloc[validation_index]
  y_validation = y.iloc[validation_index]

  # create linear regression model
  model = LinearRegression()

  #train the model
  model.fit(x_train, y_train)

  #get slope and intercept
  slope = model.coef_[0]
  intercept = model.intercept_

  #predict validation data
  y_pred = model.predict(x_validation)

  #calculate mean squared error
  mse = mean_squared_error(y_validation, y_pred)

  #store MSE
  mse_values.append(mse)

  print("========== Fold",fold,"==========")
  print("Training records: ",list(train_index))
  print("Validation records: ",list(validation_index))
  print("Slope: ", slope)
  print("Intercept: ",intercept)
  print("Regression Equation: ")
  print("y = ",intercept, " + ",slope,"x")
  print("Actual values:",y_validation.values.flatten().tolist())
  print("Predicted values:", y_pred.flatten().tolist())
  print("Validation MSE:",mse)
  print()

  fold += 1
```

```python

average_mse = average(mse_values)
print("Average MSE: ",average_mse)
```
